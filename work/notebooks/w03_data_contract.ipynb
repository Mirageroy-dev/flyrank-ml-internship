## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

```python
# Verify dataset structure, sample records, and missing values

# Display available fields
print("Fields:")
print(dataset.features)

# Load a limited number of records for analysis
records = list(dataset.take(100))

print("\nRecords analyzed:", len(records))

# Count missing values in each field
print("\nMissing values:")
missing_counts = {}

for field in dataset.features:
    count = sum(
        1 for row in records
        if row.get(field) is None
    )
    missing_counts[field] = count

print(missing_counts)

# Display date information from sample records
print("\nDate information:")
for row in records[:5]:
    print({
        "client_created_date": row.get("client_created_date"),
        "client_updated_date": row.get("client_updated_date"),
        "gsc_data_start": row.get("gsc_data_start"),
        "ga4_data_start": row.get("ga4_data_start")
    })
```

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

Data limits: The dataset is limited by missing values and uneven data coverage. In the 100-record sample, some fields had missing values, including gsc_data_start, which was missing for 36 records. The dataset also contains records with different GSC and GA4 data availability, so comparisons may not always cover the same time window. Because only a sample was analyzed from the streaming dataset, the results should be treated as observational rather than representing the complete dataset.

```python
# Verify data limitations

print("Records analyzed:", len(records))

# Count missing values
missing_counts = {}

for field in dataset.features:
    missing_counts[field] = sum(
        row[field] is None for row in records
    )

print("\nMissing values:")
print(missing_counts)

# Specifically verify gsc_data_start
print("\ngsc_data_start missing:",
      missing_counts.get("gsc_data_start"))
```
